```python
import numpy as np
import pandas as pd
```

```python
#Finding Missing Value
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
df.isna() #False-> not a null 
```

```python
df.isna().sum() #total null value in each row
```

```python
df.isna().any() # any row contain null value 
```

```python
#Removing Missing data
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df2 = pd.DataFrame(data)
df2
```

```python
df2.dropna() #drop all rows with null value
```

```python
df2.dropna(thresh=3) #remove row that has not minimum no of not null value
```

```python
#Filling Missing data
```

```python
df2
```

```python
df2.fillna(value=0) # NaN <-> 0
```

```python
replace={"A":1000,"B":2000,"C":3000,"D":4000}
df2.fillna(value=replace)
```
