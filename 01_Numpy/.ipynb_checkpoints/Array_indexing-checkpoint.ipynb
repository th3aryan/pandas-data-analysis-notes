```python
import numpy as np
```

```python
# Numpy Indexing And Slicing Of vectors 
```

```python
arr=np.arange(11,21)
arr
```

```python
arr[0] # accesing index
```

```python
arr[-1]
```

```python
arr[0:4] # [start:end] start -> < end
```

```python
arr[0:4:2] # [start:end:steps]
```

```python
arr[0::2] # [start::steps] upto end
```

```python
# Numpy Indexing and Slicing Of Matrix
```

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[5][4] # acessing individual element or arr[5,4] -> same ans
```

```python
arr[1] #arr[row no.] -> prints row
```

```python
slice=arr[0:2,1:3] #arr[row:upt0 which row , col:upto which col]
slice
```

```python
arr[3:6 , 3:5]
```

```python
arr[:,1] #Acessing Col only
```

```python
# Bollean Indexing
```

```python
arr=np.arange(1,11)
arr
```

```python
bool_index=arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```

```python
# combined above 2
bool_index=arr % 2 == 0
arr = arr[bool_index]
arr
```
