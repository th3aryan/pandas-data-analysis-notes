```python
# Advance Array Manipulation(only vector)
#Stacking Array 
```

```python
a=np.array([1,2,3,4,5])
b=np.array([6,7,8,9,10])
```

```python
np.vstack((a,b)) # vertically stacked
```

```python
np.hstack((a,b)) # horizontally stacked
```

```python
np.column_stack((a,b)) #column wise stacking
```

```python
np.row_stack((a,b))
```
